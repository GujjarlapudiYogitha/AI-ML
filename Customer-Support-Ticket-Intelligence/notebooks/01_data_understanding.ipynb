```python
SELECTED_QUEUES = [
    "Billing and Payments",
    "Customer Service",
    "Product Support",
    "Technical Support",
]

df[df["queue"].isin(SELECTED_QUEUES)]["queue"].value_counts()
```

```python
#Creating EN only DataFrame

english_df = df[df["language"] == "en"].copy()
english_df.shape
```

```python
#Selection proeject required queues

project_df = english_df[english_df["queue"].isin(SELECTED_QUEUES)].copy()
project_df.shape
```

```python
project_df["queue"].value_counts()
```

```python
project_df["queue"].value_counts(normalize=True).mul(100).round(2)
```

```python
#Display 2 tickets randomly

project_df.groupby("queue").sample(n=2, random_state=42)[["subject","body","queue"]]
```
